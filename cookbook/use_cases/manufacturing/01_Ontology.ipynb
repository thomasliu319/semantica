# Machine-tool ontology

Load the company-side OWL overlay, register it as a Semantica domain template, and check competency-question coverage. This notebook does not infer a schema from noisy text; the shop model is hand-authored.

```python
import sys
from pathlib import Path

HERE = Path.cwd()
if (HERE / "ontology" / "machine_tool.ttl").exists():
    ROOT = HERE
else:
    ROOT = HERE / "cookbook" / "use_cases" / "manufacturing"
sys.path.insert(0, str(ROOT.parent))

from manufacturing import (
    load_competency_questions,
    load_ontology_dict,
    register_manufacturing_template,
)
```

```python
ontology = load_ontology_dict()
print(f"URI: {ontology.get('uri')}")
print(f"Classes: {len(ontology.get('classes') or [])}")
print(f"Properties: {len(ontology.get('properties') or [])}")
print("Class names:", sorted(item.get('name') for item in ontology.get('classes') or []))
```

```python
domains = register_manufacturing_template()
print("Domains:", domains.list_domains())
template_ont = domains.create_domain_ontology(
    "manufacturing",
    uri="https://company.local/ontology/machine-tool/",
)
print(template_ont["name"], len(template_ont["classes"]), "classes")
```

```python
from semantica.ontology.competency_questions import CompetencyQuestionsManager

manager = CompetencyQuestionsManager()
for item in load_competency_questions():
    manager.add_question(item["question"], category=item["category"], priority=item["priority"])
report = manager.generate_report(ontology)
print(
    f"Answerable {report['validation']['answerable']}/{report['validation']['total_questions']}"
)
for question in report["questions"]:
    print("-", question["answerable"], question["question"])
```
